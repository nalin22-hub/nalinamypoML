```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ==========================================
# 1. LOAD & INSPECT DATASET
# ==========================================
data = {
    "Study_Hours": [2, 3, 5, 1, 6, 4, 2, 7, 3, 8, 5, 1],
    "Attendance": [60, 65, 75, 40, 85, 70, 50, 90, 55, 95, 80, 35],
    "Previous_Marks": [45, 50, 60, 30, 70, 55, 35, 80, 40, 85, 65, 25],
    "Final_Marks": [48, 52, 65, 35, 78, 60, 40, 88, 45, 92, 72, 30],
    "Result": ["Fail", "Pass", "Pass", "Fail", "Pass", "Pass",
               "Fail", "Pass", "Fail", "Pass", "Pass", "Fail"]
}

df = pd.DataFrame(data)

print("=== DATASET PREVIEW ===")
print(df.head())
print("\n=== DATASET INFO & MISSING VALUES ===")
print(df.info())
print("\nMissing values:\n", df.isnull().sum())

# Common input features (X)
features = ["Study_Hours", "Attendance", "Previous_Marks"]
X = df[features]

# ==========================================
# 2. TASK 1: REGRESSION MODEL (FINAL MARKS)
# ==========================================
y_reg = df["Final_Marks"]

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X, y_reg, test_size=0.25, random_state=42
)

reg_model = LinearRegression()
reg_model.fit(X_train_reg, y_train_reg)

y_pred_reg = reg_model.predict(X_test_reg)

mae = mean_absolute_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mean_squared_error(y_test_reg, y_pred_reg))
r2 = r2_score(y_test_reg, y_pred_reg)

print("\n=== TASK 1: LINEAR REGRESSION EVALUATION ===")
print(f"MAE  : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R2   : {r2:.4f}")

# Predict Final Marks for a new student
new_student = pd.DataFrame({
    "Study_Hours": [6],
    "Attendance": [82],
    "Previous_Marks": [68]
})

predicted_marks = reg_model.predict(new_student)[0]
print(f"\nNew Student Predicted Final Marks: {predicted_marks:.2f}")

# ==========================================
# 3. TASK 2: CLASSIFICATION MODEL (PASS/FAIL)
# ==========================================
# Map Pass -> 1, Fail -> 0
label_map = {"Fail": 0, "Pass": 1}
inv_label_map = {0: "Fail", 1: "Pass"}
y_clf = df["Result"].map(label_map)

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X, y_clf, test_size=0.25, random_state=42, stratify=y_clf
)

clf_model = LogisticRegression(random_state=42)
clf_model.fit(X_train_clf, y_train_clf)

y_pred_clf = clf_model.predict(X_test_clf)

acc = accuracy_score(y_test_clf, y_pred_clf)

print("\n=== TASK 2: LOGISTIC REGRESSION EVALUATION ===")
print(f"Accuracy: {acc * 100:.2f}%")
print("\nConfusion Matrix:\n", confusion_matrix(y_test_clf, y_pred_clf))
print("\nClassification Report:\n", classification_report(
    y_test_clf, y_pred_clf, target_names=["Fail", "Pass"], zero_division=0
))

# Predict Pass/Fail for the same new student
pred_class_num = clf_model.predict(new_student)[0]
pred_class_prob = clf_model.predict_proba(new_student)[0][1]
print(f"New Student Predicted Result: {inv_label_map[pred_class_num]} (Pass Prob: {pred_class_prob:.2f})")
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ==========================================
# 1. LOAD & INSPECT DATASET
# ==========================================
data = {
    "Study_Hours": [2, 3, 5, 1, 6, 4, 2, 7, 3, 8, 5, 1],
    "Attendance": [60, 65, 75, 40, 85, 70, 50, 90, 55, 95, 80, 35],
    "Previous_Marks": [45, 50, 60, 30, 70, 55, 35, 80, 40, 85, 65, 25],
    "Final_Marks": [48, 52, 65, 35, 78, 60, 40, 88, 45, 92, 72, 30],
    "Result": ["Fail", "Pass", "Pass", "Fail", "Pass", "Pass",
               "Fail", "Pass", "Fail", "Pass", "Pass", "Fail"]
}

df = pd.DataFrame(data)

print("=== DATASET PREVIEW ===")
print(df.head())
print("\n=== DATASET INFO & MISSING VALUES ===")
print(df.info())
print("\nMissing values:\n", df.isnull().sum())

# Common input features (X)
features = ["Study_Hours", "Attendance", "Previous_Marks"]
X = df[features]

# ==========================================
# 2. TASK 1: REGRESSION MODEL (FINAL MARKS)
# ==========================================
y_reg = df["Final_Marks"]

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X, y_reg, test_size=0.25, random_state=42
)

reg_model = LinearRegression()
reg_model.fit(X_train_reg, y_train_reg)

y_pred_reg = reg_model.predict(X_test_reg)

mae = mean_absolute_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mean_squared_error(y_test_reg, y_pred_reg))
r2 = r2_score(y_test_reg, y_pred_reg)

print("\n=== TASK 1: LINEAR REGRESSION EVALUATION ===")
print(f"MAE  : {mae:.2f}")
print(f"RMSE : {rmse:.2f}")
print(f"R2   : {r2:.4f}")

# Predict Final Marks for a new student
new_student = pd.DataFrame({
    "Study_Hours": [6],
    "Attendance": [82],
    "Previous_Marks": [68]
})

predicted_marks = reg_model.predict(new_student)[0]
print(f"\nNew Student Predicted Final Marks: {predicted_marks:.2f}")

# ==========================================
# 3. TASK 2: CLASSIFICATION MODEL (PASS/FAIL)
# ==========================================
# Map Pass -> 1, Fail -> 0
label_map = {"Fail": 0, "Pass": 1}
inv_label_map = {0: "Fail", 1: "Pass"}
y_clf = df["Result"].map(label_map)

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X, y_clf, test_size=0.25, random_state=42, stratify=y_clf
)

clf_model = LogisticRegression(random_state=42)
clf_model.fit(X_train_clf, y_train_clf)

y_pred_clf = clf_model.predict(X_test_clf)

acc = accuracy_score(y_test_clf, y_pred_clf)

print("\n=== TASK 2: LOGISTIC REGRESSION EVALUATION ===")
print(f"Accuracy: {acc * 100:.2f}%")
print("\nConfusion Matrix:\n", confusion_matrix(y_test_clf, y_pred_clf))
print("\nClassification Report:\n", classification_report(
    y_test_clf, y_pred_clf, target_names=["Fail", "Pass"], zero_division=0
))

# Predict Pass/Fail for the same new student
pred_class_num = clf_model.predict(new_student)[0]
pred_class_prob = clf_model.predict_proba(new_student)[0][1]
print(f"New Student Predicted Result: {inv_label_map[pred_class_num]} (Pass Prob: {pred_class_prob:.2f})")
# ==========================================
# 4. INTERACTIVE USER INPUT & PREDICTION
# ==========================================
print("=== ENTER NEW STUDENT DETAILS FOR PREDICTION ===")

try:
    study_hours = float(input("Enter Study Hours per day (e.g., 5): "))
    attendance = float(input("Enter Attendance Percentage (0 - 100): "))
    previous_marks = float(input("Enter Previous Exam Marks (0 - 100): "))

    # Create DataFrame matching training feature names
    user_student_df = pd.DataFrame({
        "Study_Hours": [study_hours],
        "Attendance": [attendance],
        "Previous_Marks": [previous_marks]
    })

    # 1. Regression Prediction (Final Marks)
    pred_marks = reg_model.predict(user_student_df)[0]
    # Clip marks between 0 and 100 for realistic display
    pred_marks_clipped = np.clip(pred_marks, 0, 100)

    # 2. Classification Prediction (Pass / Fail)
    pred_class = clf_model.predict(user_student_df)[0]
    pred_prob = clf_model.predict_proba(user_student_df)[0]

    result_label = inv_label_map[pred_class]
    confidence = pred_prob[pred_class] * 100

    print("\n" + "=" * 45)
    print("       STUDENT PERFORMANCE PREDICTION")
    print("=" * 45)
    print(f"Input -> Study Hours: {study_hours} | Attendance: {attendance}% | Prev Marks: {previous_marks}")
    print("-" * 45)
    print(f"Predicted Final Marks : {pred_marks_clipped:.2f} / 100")
    print(f"Predicted Result      : {result_label} (Confidence: {confidence:.2f}%)")
    print("=" * 45)

except ValueError:
    print("\nInvalid input! Please enter numeric values only (e.g., 4.5, 80, 65).")
```
